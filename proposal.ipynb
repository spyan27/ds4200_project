```python
import pandas as pd 
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px


# ----- need to convert yearly data into a better way 
```

```python
# load in data
df = pd.read_csv("medicare_with_conditions.csv")
df.shape
```

```python
# filter & focus on 2023 data & overall data  
outlier_23_df = df[df['Outlier_Flag_2023'] == 1]
outlier_23_df
```

```python
outlier_23_df.columns
```

```python
fig = px.scatter(outlier_23_df[outlier_23_df['Mftr_Name'] == 'Overall'], x='Avg_Spnd_Per_Clm_2023', y='Tot_Spndng_2023',
                 hover_data=['Brnd_Name', 'Gnrc_Name', 'Mftr_Name'])

fig.update_layout(
    title='Average Spend per Claim vs Total Spending in 2023 (Overall)',
)

fig.show()
```

```python
# stacked bar chart to show brand vs generic name
overall_df = df[df['Mftr_Name'] == 'Overall']

# create a new column for brand vs generic drug
drug_types = []
for i in range(len(overall_df)):
    if overall_df.iloc[i]['Brnd_Name'] != overall_df.iloc[i]['Gnrc_Name']:
        drug_types.append('Brand')
    else:
        drug_types.append('Generic')

overall_df['Drug_Types'] = drug_types

# find total spending by year and drug type (brand vs generic)
lst = []
for year in [2019, 2020, 2021, 2022, 2023]:
    total_spending_year = 'Tot_Spndng_' + str(year)
    
    brand_total = overall_df[overall_df['Drug_Types'] == 'Brand'][total_spending_year].sum()
    generic_total = overall_df[overall_df['Drug_Types'] == 'Generic'][total_spending_year].sum()
    
    lst.append({'Year': year, 'Drug_Types': 'Brand', 'Total Spending': brand_total})
    lst.append({'Year': year, 'Drug_Types': 'Generic', 'Total Spending': generic_total})

plot_df = pd.DataFrame(lst)

fig = px.bar(plot_df, x='Year', y='Total Spending', color='Drug_Types', barmode='stack')

fig.update_layout(title='Total Medicare Spending Each Year by Brand vs Generic Name From 2019-2023')

fig.show()
```

```python
#first remove NaN values
outlier_23_df = outlier_23_df.dropna(subset=['Mftr_Name']) # remove NaN values in Mftr_Name column
outlier_23_df = outlier_23_df[outlier_23_df['Mftr_Name'] != 'Overall'] # remove rows where Mftr_Name is 'Overall'
outlier_23_df
```

```python
# average growth rate of drug spending per year by Manufacturer 2019-2023
df['Spndng_Growth_2023'] = (df['Tot_Spndng_2023'] - df['Tot_Spndng_2022']) / df['Tot_Spndng_2022']
df['Spndng_Growth_2022'] = (df['Tot_Spndng_2022'] - df['Tot_Spndng_2021']) / df['Tot_Spndng_2021']
df['Spndng_Growth_2021'] = (df['Tot_Spndng_2021'] - df['Tot_Spndng_2020']) / df['Tot_Spndng_2020']
df['Spndng_Growth_2020'] = (df['Tot_Spndng_2020'] - df['Tot_Spndng_2019']) / df['Tot_Spndng_2019']
df['Avg_Spndng_Growth'] = df[['Spndng_Growth_2023', 'Spndng_Growth_2022', 'Spndng_Growth_2021', 'Spndng_Growth_2020']].mean(axis=1)

# Sort the dataframe by Manufacturer Name
df = df.sort_values(by='Mftr_Name')


fig = px.bar(df, 
             x='Mftr_Name',  
             y='Avg_Spndng_Growth', 
             hover_data=['Brnd_Name', 'Gnrc_Name', 'Mftr_Name'])

fig.update_layout(
    xaxis=dict(
        title='Manufacturer Name',
        tickangle=-45,
        automargin=True,
        rangeslider=dict(visible=True)  # Add a range slider for scrollable x-axis
    ),
    height=1000,
    width=2000,
    xaxis_title='Manufacturer Name',
    yaxis_title='Average Spending Growth Rate'
)
fig.update_layout(
    title='Average Spending Growth Rate (2020-2023) by Manufacturer',
    height=1000,
    width=2000,
)
fig.show()
```

```python
#filter to "overall" manufacturer for pricing analysis
overall_df = df[df['Mftr_Name'] == 'Overall']
overall_df
```

```python
# melt the dataframe to long format by year for “Avg_Spnd_Per_Dsg_Unt_Wghtd_YYYY”
price_columns = [col for col in overall_df.columns if col.startswith('Avg_Spnd_Per_Dsg_Unt_Wghtd_')]

df_long = overall_df.melt(
    id_vars=['Brnd_Name', 'Gnrc_Name', 'Disease_Category'],
    value_vars=price_columns,
    var_name='Year_Column',
    value_name='Avg_Price'
)

df_long
```

```python
#extract Year from 'Year_Column' Name
df_long['Year'] = df_long['Year_Column'].str.extract(r'(\d{4})').astype(int)
#drop old 'Year_Column'
df_long = df_long.drop(columns=['Year_Column'])
df_long
```

```python
#remove missing values where dugs don't have pricing data
df_long = df_long.dropna(subset=['Avg_Price'])
df_long
```

```python
# compute yearly average price for D3 plot on Average Price Trends per Year (2019-2023)
yearly_avg_price_trends = df_long.groupby('Year')['Avg_Price'].mean().reset_index()
yearly_avg_price_trends
```

```python
# export to csv for d3 plot
yearly_avg_price_trends.to_csv("yearly_avg_price_trends.csv", index=False)
```
